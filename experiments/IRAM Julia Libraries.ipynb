# Numerical Comparison of Restart Methods of the Arnoldi Method

## Imports

```julia
using LinearAlgebra
using LinearMaps
using MatrixDepot
using ArnoldiMethod
using Arpack
```

## Julia Libraries

We check what final accuracy the implementations of Krylov-Schur and IRAM of Julia libraries achieve.


```julia
A = matrixdepot(r"rajat12") + 10 * I
n = size(A, 1)
num_eig = 4

display(A)
```

### Krylov-Schur

```julia
decomp, history = partialschur(A, nev=num_eig, tol=1e-15, which=:LM, restarts=50, mindim=15, maxdim=30);

λs, X = partialeigen(decomp)

println("Maximum residual: ", maximum(abs.(A * X - X * Diagonal(λs))))
```

### IRAM

```julia
evals, evecs = eigs(A, nev=4, which=:LM, maxiter=50, ncv=30, tol=1e-15)

println("Maximum residual: ", maximum(abs.(A * evecs - evecs * diagm(evals))))
```
